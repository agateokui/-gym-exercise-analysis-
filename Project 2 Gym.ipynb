```python
from datascience import *
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plots
```

```python
gym= Table.read_table('megaGymDataset.csv')
gym
```

```python
exercises = gym.select("Title", "BodyPart", "Level", "Rating", "RatingDesc") 
exercises
```

```python
body_part_abdominal= exercises.where('BodyPart', are.containing('Abdominals'))
body_part_abdominal
```

```python
body_part_abdominal_beg=body_part_abdominal.where('Level', are.containing('Beginner'))
body_part_abdominal_beg
```

```python
body_part_abdominal_beg_fl= body_part_abdominal_beg.where('Rating', are.above(6.9))
body_part_abdominal_beg_fl
```

```python
body_part_abdominal_beg_fl.show(16)
```
